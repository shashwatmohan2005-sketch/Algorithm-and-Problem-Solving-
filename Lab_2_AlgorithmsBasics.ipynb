```python
# Exercise 1
import timeit
def func1(a, b):
  result = 0
  for i in range(0, b):
    result += a
  return result
```

```python
# Exercise 4
import timeit
def func4(a, b):
    b.sort() # Time complexity for this function is O(Nlog N) where N = len(b)
    intersect = 0
    for x in a:
        if binary_search(b, x) >= 0:
            intersect = intersect + 1
    return intersect

def binary_search(arr, target):
    low, high = 0, len(arr) - 1
    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1
```

```python
# Sample function call: func4([1, 2, 3, 4], [2, 4, 6, 8])
x=print(func4([1, 2, 3, 4],[2, 4, 6, 8]))
start_time = timeit.default_timer()
x=print(func4([1, 2, 3, 4],[2, 4, 6, 8]))
print('>> elapsed time', (timeit.default_timer() - start_time) * 1000, 'ms')
# This function counts how many elements from list 'a' are also present in list 'b' using sorting and binary search

# This function performs a binary search on a sorted list and returns the index of the target if found, otherwise -1
```

```python
# Exercise 5
import timeit
def func5(n):
    guess = 1
    while guess * guess <= n:
        if guess * guess == n:
            return guess
        guess = guess + 1
    return None
```

```python
# Exercise 6
import timeit
def func8(A):
    func8_helper(A, 0, len(A) -1)

def func8_helper(A, i, j):
    #Input: An array A and two indices i and j.
    n = j - 1 + 1
    if n ==2:
        if A[i] > A[j]:
            # Swap A[i] and A[j]
            A[i], A[j] = A[j], A[i]
        elif n > 2:
            m = n // 3
            func8_helper(A, i, j - m)
            func8_helper(A, i + m, j)
            func8_helper(A, i, j - m)
        return A
```

```python
# Exercise 7
import timeit
def func12(arr, k):
    arr_len = len(arr)
    k = k % arr_len
    # Initialize a new array of the same length as 'arr'
    custom_arr = [0] * arr_len

    for i in range(arr_len):
        mystery_index = (i + k) % arr_len
        custom_arr[mystery_index] = arr[i]
    return custom_arr

```

```python
#write a function named as max_min to find minimum (or maximum) element in an array
def max_min(arr, find='max'):
    if not arr:
        return None
    result = arr[0]
    for num in arr:
        if find == 'max' and num > result:
            result = num
        elif find == 'min' and num < result:
            result = num
    return result
numbers = [12, 3, 4, 15,45,78,450]
print(max_min(numbers, 'max'))
print(max_min(numbers, 'min'))
```
